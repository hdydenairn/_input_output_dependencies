```python
import pymrio
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Load the original EXIOBASE data
exio3 = pymrio.parse_exiobase3(
    path=r"C:\Users\Dyde-nairn\Documents\python\_input_output_dependencies\IOT_2022_pxp.v3.8.2.zip"
)
```

```python

# Compute Leontief inverse manually
exio3.L = pymrio.calc_L(exio3.A)
```

```python
# Step 1: Original final demand vector
original_vector = exio3.Y.sum(axis=1).copy()

# Step 2: Create shocked vector with 1% increase in French wind electricity
shock_vector = original_vector.copy()
shock_vector[("FR", "Electricity by wind")] *= 1.01

# Step 3: Combine into a DataFrame for comparison
comparison_df = pd.DataFrame({
    "Original Final Demand": original_vector,
    "Shocked Final Demand": shock_vector,
    "Difference": shock_vector - original_vector
})

# Step 4: Sort by difference to highlight the shock
comparison_df_sorted = comparison_df.sort_values(by="Difference", ascending=False)

# Step 5: Display the top rows
print(comparison_df_sorted.head(10))
```

```python

# Step 1: Original final demand vector
original_vector = exio3.Y.sum(axis=1).copy()

# Step 2: Shocked final demand vector (1% increase for French wind electricity)
shocked_vector = original_vector.copy()
shocked_vector[("FR", "Electricity by wind")] *= 1.01

# Step 3: Compute total output before and after the shock
original_output = exio3.L @ original_vector
shocked_output = exio3.L @ shocked_vector

# Step 4: Compute the difference in output
delta_output = shocked_output - original_output

# Step 5: Convert to DataFrame for sorting
delta_output_df = delta_output.reset_index()
delta_output_df.columns = ["Country", "Industry", "Output Change"]

# Step 6: Sort and select top 30 affected industries
top_30_affected = delta_output_df.sort_values(by="Output Change", ascending=False).head(30)

# Step 7: Plot the results
plt.figure(figsize=(12, 8))
plt.barh(top_30_affected["Country"] + " - " + top_30_affected["Industry"], top_30_affected["Output Change"], color='skyblue')
plt.title("Top 30 Industries Affected by 1% Increase in Final Demand for French Wind Electricity")
plt.xlabel("Increase in Output (EUR)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Step 2: Create shock vectors with a 1% increase for wind and solar
shock_vector_wind = pd.Series(0.0, index=exio3.L.columns)
shock_vector_wind[("FR", "Electricity by wind")] = original_demand * 0.01  # 1% increase

shock_vector_solar = pd.Series(0.0, index=exio3.L.columns)
shock_vector_solar[("FR", "Electricity by solar photovoltaic")] = original_demand * 0.01  # 1% increase

# Step 3: Calculate the change in output using the Leontief inverse
delta_output_wind = exio3.L.dot(shock_vector_wind)
delta_output_solar = exio3.L.dot(shock_vector_solar)

# Step 4: Convert to DataFrames for filtering
df_wind = delta_output_wind.reset_index()
df_wind.columns = ["Country", "Industry", "Output_Change_Wind"]

df_solar = delta_output_solar.reset_index()
df_solar.columns = ["Country", "Industry", "Output_Change_Solar"]

# Step 5: Filter for Aluminium sector
sector_name = "Aluminium ores and concentrates"
impact_wind = df_wind[df_wind["Industry"] == sector_name]
impact_solar = df_solar[df_solar["Industry"] == sector_name]

# Step 6: Extract aluminium production
idx = pd.IndexSlice
aluminium_output = exio3.x.loc[idx[:, sector_name], :]
aluminium_output_df = aluminium_output.reset_index()
aluminium_output_df.columns = ["Country", "Industry", "Aluminium_Production"]

# Merge all data
merged_df = aluminium_output_df.merge(impact_wind, on=["Country", "Industry"])
merged_df = merged_df.merge(impact_solar, on=["Country", "Industry"])

# Sort by aluminium production (highest to lowest)
sorted_df = merged_df.sort_values(by="Aluminium_Production", ascending=False).reset_index(drop=True)

# --- Plot two-panel figure ---
fig, axes = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

bar_width = 0.35
positions = sorted_df.index

# Top: Aluminium production (black bars, width 0.35)
axes[0].bar(positions, sorted_df["Aluminium_Production"], width=bar_width, color="black", label="Aluminium Production")
axes[0].set_title("Aluminium Ore Production by Country (Pre-Shock)")
axes[0].set_xlabel("Producing Region")
axes[0].set_ylabel("Aluminium Ore Production (Million EUR)")
axes[0].legend()

# Bottom: Impact from two shocks (blue and red grouped bars)
axes[1].bar(positions - bar_width/2, sorted_df["Output_Change_Wind"], width=bar_width, color="#00abfb", label="Wind Energy")
axes[1].bar(positions + bar_width/2, sorted_df["Output_Change_Solar"], width=bar_width, color="#e85f3b", label="Solar Energy")

axes[1].set_title("Impact on Country Specific Aluminium Ore Production from 1% Demand Shocks to the Wind and Solar Energy Sectors in France (Applied Separately)")
axes[1].set_ylabel("Increase in Aluminium Ore Output (Million EUR)")
axes[1].set_xlabel("Country Specific Response to Shock")
axes[1].legend()

# Shared xticks with EXIOBASE country codes
xticks_labels = sorted_df["Country"]
for ax in axes:
    ax.set_xticks(positions)
    ax.set_xticklabels(xticks_labels, rotation=90, ha='center')
    ax.tick_params(axis='x', which='both', labelbottom=True)

plt.tight_layout()
plt.savefig("aluminium_production_vs_wind_and_solar_shock.png", dpi=300)
plt.show()
```
